```python
def prime(n: int) -> bool:
    if n < 2:
        return False
    Isprime = True
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            Isprime = False
            break
    return Isprime


def prime_product(m: int) -> bool:
    if m <= 0:
        return False

    for i in range(2, m//2 + 1):
        if prime(i) and m%i == 0 and prime(m//i):
            return True

    return False

n = int(input())
print(prime_product(n))
```

```python
def del_char(s,c : str)-> str:
    if len(c)>1:
        return s
    a = list(s)
    [a.remove(i) for i in a if i == c]
    return "".join(a)

s = input()
c = input()
print(del_char(s,c))
```
