### GCD calcualation using Recursive function 


```python
def gcd(m,n):
    (a,b) = (max(m,n), min(m,n))
    if a%b == 0:
        return (b)
    else:
        return (gcd(b, (a-b)))
print(gcd(16,20))
```

## Euclid's Algorithm
- Suppose `n` does not divide `m`
- Then `m = qn + r`
- Suppose `d` divides both `m` and `n`
- Then `m = ad`, `n = bd`
- `m = qn + r` -- `ad = q(bd) + r`
- `r` must be of the form `cd`

### **Euclid's algorithm**
