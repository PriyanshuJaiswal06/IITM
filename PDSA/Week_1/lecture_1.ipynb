```python
def prime(n):
    (result, i) = (True, 2)
    while (result and (i<n)):
        if (n%i) == 0:
            result = False
        i += 1
    return result

def primedif(n):
    last_prime = 2
    pd = {} #Dictionary for prime difference
    for i in range(3, n+1):
        if prime(i):
            d = i - last_prime
            last_prime = i
        if d in pd.keys():
            pd[d] += 1
        else:
            pd[d] = 1
    return(pd)

print(primedif(100))
```

### Same program using exception handling


```python
def primedif(n):
    last_prime = 2
    pd = {} #Dictionary for prime difference
    for i in range(3, n+1):
        if prime(i):
            d = i - last_prime
            last_prime = i
        try:
            pd[d] += 1
        except KeyError:
            pd[d] = 1
    return(pd)
print(primedif(100))
```
