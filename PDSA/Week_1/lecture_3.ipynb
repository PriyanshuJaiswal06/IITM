## 2D Points

* A point has coordinates (x, y)
* `_init_()` initializes internal values `x, y`

* First parameter is always _self_, Here, by default a point is at (0, 0)

* Translation: shift a point by `(Ax, Ay)`
 * `(x,y) -> (x+Ax,y+Ay)`

* Distance from the origin
 `d= root(x^2+y^2)`

```python
class Point:
    def __init__(self, a=0,b=0) -> None:
        self.x = a
        self.y = b

    def translate(self, deltax, deltay):
        self.x += deltax
        self.y += deltay

    def odistance(self):
        import math
        d = math.sqrt(self.x**2 + self.y**2)
        return d

    def __str__(self):
            return('('+str(self.x)+','+str(self.y)+')')

    def __add__(self, p):
        return(Point(self.x + p.x, self.y + p.y))
    
point1 = Point()
point1.translate(3,4)
print(point1.odistance())
```

```python
p = Point(3, 4)
q = Point(7, 10)
```

```python
print(p + q)
```

```python
p.odistance(), q.odistance()
```

### Polar Coordinates

* `(r, 0)` instead of `(x, y)`
 * `r = sqrt(x**2 + y**2)`
 * `0 = tan^-1(y/x)`

- Distance from origin is just `r`
- Translation
  - Convert `(r, 0)` to `(x, y)`
  - `x = rcos0, y = r sin 0`
  - Recompute `r, 0` from `(x+Ax,y+Ay)`

* Interface has not changed
 - User need not be aware whether representation is `(x, y)` or `(r, 0)`


```python
import math
class Point:
    def __init__(self, a=0, b=0):
        self.r = math.sqrt(a**2 + b**2)
        if a == 0:
            self.theta = math.pi/2
        else:
            self.theta = math.atan(b/a)

    def odistance(self):
        return (self.r)

    def translate(self, deltax, deltay):
        x = self.r*math.cos(self.theta)
        y = self.r*math.sin(self.theta)
        x += deltax
        y += deltay
        self.r = math.sqrt(x**2 + y**2)
        if x == 0:
            self.theta = math.pi/2
        else:
            self.theta = math.atan(y/x)
        return self.theta

    
point1 = Point()
print(point1.odistance())
print(point1.translate(3,4))
print(point1.odistance())
```
