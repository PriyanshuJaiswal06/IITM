```python
def matrixMultiply(A,B:list):
    (m,n,p)= (len(A), len(B),len(B[0])) #len(B[0]) here finds the number of columns in second matrix

    C = [[0 for i in range(p)] for j in range(m)]  #The resultant matrix, where every value is initilized to 0.

    for i in range(m):
        for j in range(p):
            for k in range(n):
                C[i][j] = C[i][j] + A[i][k]*B[k][j]

    return (C)

A = [[1,2,3,4],[5,6,7,8]]
B = [[1,2],[3,4],[4,5],[8,8]]
print(matrixMultiply(A,B))

```

```python

def numberOfBits(n):
    count = 1
    while n>1:
        count += 1
        n = n//2
    return(count)

print(numberOfBits(99))
```
