```python
# type solution code here
def L1(List:list)->list:
    result = [List[0]]

    for i in range(1, len(List)):
        if List[i][0] >= result[-1][0]:
            result.append(List[i])
        else:
            result.append(result[-1])
            # print(result)
            for j in range(len(result)-2,-1,-1):
                if j == 0:
                    result[j] = List[i]
                elif List[i][0] < result[j-1][0]:
                    result[j] = result[j-1]
                else:
                    result[j] = List[i]
                    break
    return result

def L2(L:list)->list:
    for i in range(len(L)):
        for j in range(len(L)):
            if L[i][0] == L[j][0]:
                if int(L[i][1:3]) > int(L[j][1:3]):
                    L[i], L[j] = L[j], L[i]
    return L

def combinationSort(strList:list):
    list1 = L1(strList)
    lcr = []
    lcr.extend(list1)
    L2(lcr)
    return (list1, lcr)

print(combinationSort(['d34', 'g54', 'd12', 'b87', 'g1', 'c65', 'g40', 'g5', 'd77']))
# print(L1([))
# print(L2(L1(['d34', 'g54', 'd12', 'b87', 'g1', 'c65', 'g40', 'g5', 'd77'])))
```
